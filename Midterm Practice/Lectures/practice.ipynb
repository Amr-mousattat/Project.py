```python
measurements = [12.4, 1.0, 13.2, 0.0, 14.1]

def calculateList(measurements: list):
    size = 0
    total = 0
    for x in measurements:
        if(x < 0): continue
        if(x == 0): break
        size += 1
        total += x
    return size, total
print(calculateList(measurements))
```

```python
"""
calculate:

mean;
minimum;
maximum;
range (max - min);
a new list containing each measurement minus the mean.
"""
measurements = [12.4, 13.1, 11.8, 14.0, 12.7]

def decipher(measurements: list):
    mean = sum(measurements)/len(measurements)
    minimum = min(measurements)
    maximum = max(measurements)
    range = maximum - minimum
    shifted = [x-mean for x in measurements]
    return mean, minimum, maximum, range, shifted

print(decipher(measurements=measurements))
```

```python
#Part 4 - Middle Cross Pattern (9 marks)
"""
Write a Python program that asks the user to enter a positive odd integer n.
Using nested loops, display the pattern shown below.
Enter a positive odd integer: 5
- - + - -
- - + - -
+ + + + +
- - + - -
- - + - -
"""
def drawCross(oddInteger):
    result = ""
    #Nested loop
    for i in range(oddInteger): 
        for j in range(oddInteger):
            if(i == oddInteger//2 or j == oddInteger//2):
                result += "+ "
            else:
                result += "- "
        result += "\n"
        i+=1
        j+=1
    return result
        

number = int(input("enter positive odd integer: "))
print(drawCross(number))
```

```python
def adjust_measurements(values):
    result = []
    for x in values:
        if x <= 10:
            result.append(10)
        elif(x <= 20):
            result.append(x)
        else: result.append(20)
    return result

print(adjust_measurements([7, 10, 15, 23]))
```

```python
def alternating_total(limit):
    number = 1
    total = 0
    while number <= limit:
        if(number%2 == 0):
            total -= number
        else: total += number
        if(abs(total) > 10):
            break
        number += 1
    return total

print(alternating_total(10))
```
